```python
import base64
import hashlib
import json
import pathlib
import zipfile
import zlib

SHARD_ID = "shard_0001"
SOURCE_DATASET = "hungluongduc/hung-dev-shard0001-44b8532-20261027-v2/versions/1"
PAYLOAD_DATASET = "hungluongduc/hung-matched-payload-44b8532-20261027-v1/versions/1"
SAVED_TRAINING_VERSION = 362612528
INDEPENDENT_READBACK_VERSION = 362633606
PAYLOAD_SHA256 = "23a6d420523d990d07ff07d814ed642a851925283a102a3ae806f3b666e1d6f4"
ARCHIVE_SHA256 = "29e3db51106402942959f80762e4c27995ad4ec9f2253fd96cc8a4681d3b2dcd"
ARCHIVE_BYTES = 65931034
PROVIDER_PLAN_SHA256 = "80a1666cf2083594c9a2bc0b0a4dc3e30118eda43d07187cd5b6d8b51f90e0bb"
PROVIDER_PARTITION_SHA256 = "e2c6cb6f15a7473c6cdfd7912fefa838d63a770cf352dff93c2cccc3ae86a15c"
VERIFIER_SHA256 = "75b80532ad5d7149f461a6260c4a3c630a03007059a98c2790e9d878b085cf10"
VERIFIER_B85 = "c-pl*+j85+lJESAS?oi|1fnHbvLsfRDsAkX<Fkq5vc0L=!)2jBU?@Q*0U7{hIa>Yu_GJb$0}Mz>C+i0bm`it0Pv54e!5|1e)}@TsqR6GF;#--D*YZ}bip^T)wWzM+G8I`~7a|tbI$o_pk-+#`9u)cNo7l+m;HFrJO<5Gn@nG=jI;+I3EVE@MOOY0`5_wUJa+`~~jI%t;uR^hub#g8KGrqc7Ns-2NEQ(UZ+xi+XW(mL+2_OQ<)poO4WilNA#40Ot@%D8!7N4&50I}XywOAKhpcG~cc^4<OT1By~H``joIg+2sjfDU58c1Hi>maL%Oah8S;xf6;ZlzcPfwJ6`8A2=wC@k$ZOJ(_8C1By{B2GTTLsFFKcn}1^V6ZHUwTPnSw%(R9ibS^F6eVJYmB%$;s0IVwR$gu5vXc7!I<Br)*+ReGR7I|zH*sxu{h4i+*-A39`r8KR)QB$gZ-c>azrTL-YxFbxFQy_mIa!>azC6A-y-3eqEf>d!uP%<x&yP-DUMw#zj*|G~<l_A0+3{I0cz^lf&AU(0`(H2LMIZluiIb1hlaurG`1Jf_`Ra5LFVE!R#mUL3JUn{2I5|Fl`SR#wad<A{<0O&ir>DuQ^V36lu}qGR7xDR-lm_VKhfi-mz5Ts`e3qs!PYxI7C-Lb~yh!4UG+7+W^Oq+Vhw0(@d2)DmbauX6F4D!~B0WD!U!5NxpQf)~oyNy!7s=TYm_114QXnTH5Y6M$S3~5}M00EW9{!HR!B3*9%gI2%zcjm&AS6@G>#>?QqTa7reNB_c1xUy+C>Oy<#1&2lrC<p7ZwV5bTyOKwAWR~wWjS2M>qQz*)W|W4#_;Iy_(Xg!@Xsg|iy#;o2!@vNb^}5$hXjwwC?&{FuKTa$9X|n)yx_H5FJ!5gtb}!<g<|~AZ1W5J4MiTWC6PotsJEMyoPj8Z*ff6(KdS6c34Mo#ISLu`2jTxe@`tehFcm96WpU@R?8E6{KNCy&Wt){SvB=X5^>QK>MF9qFEvqWNvKaRcj3HB&Ey;eFDGfnj%_=GWh*w+rrYwtcsGt~ZaT=jQ_ep4)CG`w2hN!@5b8ZnF@E_-mS4EswG@G~=)$*<$$~-AjQ2uGKt(OPq!ANs}*OpaTUe$4)$RWWB2@NoN8N9jM$fO3#_>YgjzZ1pc2Hqy(9^pR(4j*ZzS_T-JT*5+%MKz=`n^g>qbBPaOb!Dx1+U_D35@ji|cVG!74rKcD|EsK`sdL2+#0uQFhrhERON03Yn@HI3FNhjaGYfEBFn^#qiCnE{5<ciz_{=hj56wIp+e|F-Wp>3=>9LtiZ#<nJ4FH~_LFipT`9+#WV#>m7lL?y<yaooar^9uCSJNQXMi9@Clo>l<#I0Si;q@%walstrl%APpM6208?@-2*h_)jXjNh^fq~_{H+7QL4?llN49vP9u(`ScsS`c&t{tk{THH)ugP46S}w1?tH(Tw<UDvl3<A8&Jzxhz$Pkx#M?T*(H9KfK9^0VEbA1pJPm0LVw0j3%mrU<LYRc143->BiY1i%F<AJwzBv;t?NE1Hl4HJ?LKvj~<CvVL`ix1RU)Oi!ek8m#bCr6~mvrYCJm-vnmpb4#d}Mu-yPpR<gWBzYg(N`At+whVjrcOGrpS5yu#AXlwgfe#5m7al>KpxE}gT3&N+htgnmIHnAWiyqfV9YzY7uiGQ0m3uzi%o2<C@OT2=xrGvaGwq+s_T@A4fZkn0Bkhlk-h?n3GKycc)0nYtH*u~EsHq{J?1_bP`0wSFtyUlxuR}G&^W8gsR(87V$1aE<AJ5kn%S;=p-tBrYTw)Qndx)3#Dn79Vvj_U$94j~DUV8Ou}#*k!10a68<53w&IZFaMi@cIoHq|FXDD6Lw6oP0*pxX#dPbKfkW<|vK7f%b(>>pH%RiZZ(bC(u&LhAj{~$|21ji45%y$H>%7NStOGD1HL1#WyD6Cs11O;@tHE*pLQxqsMAVG`<RWfU<00V{jeoEFZqqQ1tQ+#F4A?JQ{YI`a#M*Fks#+QGlYX;*HX87I^fY=q{A_&$S!Zy;Ci!5guRyBQKzU%j^zqt{OlmoC7$s)C!7R88{dffCk(Lm5jkz$EzrhK@?CEVAubYd8Ab1fjCDbDHWL`gH4Fx#ob`6yrbr}Fwef{4z%UW_=9_5+(TeD5epvxFw1>%EtAiZLzHOotp?dI<F61B;UjEFypSss5&41@gtE=EFIx#lGApYID+oLHX6(FoeiQJtO(?qH2Pd=yvk+ofxn}=|$Xa^N@`83r?TiY4mX@!jzz2pv)I-aAcySMJ+PgU<&7cG6K%9e?HBsRaw^Z9z?FMYX#ZZM;5s?~Q7scl(bE-F5V@UhMH74BkXgpWDFEx)1T7>6=YDmwto1qfm_9=n7;6L%dB%1Rk{AFuq)W*_^Nt-hPHfQl_ys1I-5KbQy&s?RIB$t~43{V$}(!CzNv(>t*hFyf^7L0PqSORn~NnF*{7`Jj247wvbqy^xpb2EeaPsxvO5zulw#8d+~bnQ@V-vaJhKiivA7DeqlhX6Q?y%a!pdDs^83KaHoM9py*#!puFUC3BayY7&qIDTY>Vz>YWy%DK(k#1LbJ~8ejz|JyT6C>b6b`@2WJvkm@ObN&IMO=a1%w?3UvcPugj+K7io;KjnPVOM3LG5zO;U=#wps2OAU{uGCB*2OR>#SO%9|wJ;&3kAwDk;_&54fD_=#R~f9ctEErRuNN#b>z_M0=ug1gS~y`IthI#woEYjPf0RsC5~J=!N<?(m09bD!z)yY7X(y$hWaKmXj+gI^59dCnNdHLUk|(A(8oZElWKAlfdG-)uDMxc;<t)PoVRc21-YcBiAi3-%#EKlc$#O;vQdZ_n@)?hD0E)c|l%5G=$+1q9pRX(P(4QtTtkH8`ONz1l&l8#RVf(!{Fcmb|72I3S@$VU^Xvs6wEi<I?}x=dxQs2XEq)(Y7jL%M!<#)c`mm`c$yrwCU9nzVvBpUR}U}nNcDnF>EC7PcXQyJ76KLeO**^Mk_l3~#$|srJHmU=t8rTF$MAnySD3LHYC#V?Xr2AAXdo`$9q8DZ$YH>xTtk@25taqo#G)L!Y66YYjdraW@PQs?i84533suA1faVqhd-NCEPk#{AVv9OPHe_m)hD_y_J!HD^Yd?}o%2YF5OCYt)Ek<L5jl}V{lsQltDmrIIc*cslxwp_6qf;sU1J&tt54dL3UVti}+S~7>%R<+!=(()PyGBqHKfz<1f{J${k-G69=uMZt)ads1M(%KO{e#%9SA!?eL}r^BLJn<jDjjw}-%<j*6WBhYixSIVMS$$Oj1y#+br@Fi;iUN84JAQoU0i^eyr*GwwibBd(2~cF3GktCw$H(1-(^_H^aK~9<yS4@>G*E4!mUMl<<dLFCDJ3Nuqc@5tgj=+qdvjiHF+Oqm)u9=?3w1lrZ=(8idmo+PA+b@fHY`T0Jv9Dx+Hn<b+*9Di@aBTyKsk|V9Zcpqy@9isiaU0GZP)uo_=#VXwmc=j9LjCyPF0Fb(Itw=|yVw1XhUIxSo&L-{!P}%_&*+qB67kO($p@6USNT&b>s7OJPTq%uR%?oq7?91pTUsXx=p$P0KDaY|exp>G6`%@Y}&Kbrv34qQtrKG+TmktvJ=lOLp_;R-W93WCh$eq4;vs<Y#FuGnirJxJ_FprfFx(2^b|XJ~avN<T}o;!1#AF+7Qm&nEM5yCIJA6p@ueMwz=WUnqWR}Z_@0G>1*eV>;(B@o8p-Wo+z-u(|Z%*hN34{9{GcpO=6POutQ(WtAF8i>qNPx?t;CMsVn@vd`|4o=d-|Tb@bRT{;{DO`TB@d>24-4Y{ju1GW{WkNfpBps3&eAp1nG8at&O40TSUqlA4eS1O0k98si?(UG=nw5&(axz~VbB5359aEWPdRY-}Shd>&{a@I@QNSDg@bBdEE_-_reh>khw#KCU!#j3n|67eK)W5_Ni@91h>r<2#}rj`WcgTgEa^)d}fEF!CRi3AS@{$1U!q<v-uPS0xGi1R1kVP6I8Vs7>IzE#-SS_ugRy%=HE1tcv1ARjjsf${%%RqjUx1ai9~uyC9*(tvV2FmeKN@5hGVsg8#~^c<HP;^qlejU8sndFYkC{2pE#pmi+uSx58k`%7*G*OT<YcH#H8zyCSmdV{?lJaS=m<E!P_i@?C@A47=R7hO1i!G)YMO&mVrakI`nJli^6&D&lQa_kuv9r(%vk-)2R65+y?EXxv<M=`ITeh=j^bvr595Z@YGdL|>V^_i|l#Jfj1tW$oxdi5ZJQd%tzPfE%>%jr@)yy91|LC*L>%gVr&hS_4cs9$+4R9WPe$K(GJc_9z&%h1UgR$B`p%9JvPo-^cfWu%r4taEKSCV!_9#(LPdWuNX*coA=DbFcjSI{liGVQ}jA=qt|`tIph2|SS>Brro1D$?cmsz6O$<>Rif&VKE-&jq>B#ADu%pwC?QKt*LP&5XGinF<82pOdz$ZL!}w(WBH)XFwkv?GRf^kydg$nZL-7<Zw}*B#IWo>L)vHrss4p$FHzrlZKVrHo1l83(_9zYs*(|hTLQ~@-ru!_3Tb0I_%~RnsN&3!STb@?D$Xug}k15^7s=c5f%dp-%ivucS^vUYlAic=^mfL|Did|4VhR*`8i+(uzyiq5Zwq7=a_n(q?&v%X6{E@~@gF!Jhpx8LgM_Pcn!5?6t3G~mTxQJ(l-N+*N2S}~niS`k{8(X`^#Ai^(ak^{M*}^s{H~S_X3su~kbTdKA;r{We@>IL(!;UfH+Y``aX3C_)lh%bar<XdsozcU*Gn*&%DUiAh@F<a)>s2EIcGNSC>1-9rS`bH0g=tq?deo?eVCrFyxdG9FWS=jT5H(jU9SnU7+G!uGg@s}lbF@rPU8u7_bp+ilL_-|)l^uH-MAa1OcD<?m{rK;`w+H`Gl?NAU>>Ae8mNYkONbDG^J}S?ywp^`MZE~*?2<0}X+B>C?Ce08$^g??Gd%G$)phtkFnqtLT|F{@S45K{Gr&Q|ZA2=FQ@v!eq^)Cps+IR8`$|)2J??|dA)1ygxRHZ6D?w|1tSJK7$Y<JQD)pNOuAa~xr?uX1Fg$q$d8BYL$O4lYzX`x>{w9)DDT8+U8A7ODkMKDzU!Z*{(!i4S97+)+$c&FTsr1X7F)ipfS>@zz}wGN`*mU)bkVt;v5=w5I3szDlQLhi9JsM%q5Le}9ITzFh<ONnW4-D!yG+5@`NB|0bG4RY~QV4%i7QDs{T`H{mRdKp;-;mtVZ<7vP%+>bY;+2v!X-~o`~NyQ}o(aiG7m3iLDrGCfZ&lhQYqj6kCScq{qq;OBS_nY24-I35TV0S0Ao|fD)P$|LH?-!Ojdu4WzKM`^^@x{IF@M*VyQ}yRMq4}W5`;ijLlPVAC!8m%}Hny|B5P@Ly0>%z`sEM-s{i$%5UB7YatjG2}mr-5~cphJzc>IT{IO>dA6;~1O3_Vd8+q*R|q#a8`Q}9HCqETY@zt=T2gQgu5jwt8N2JVKyQL`Vw?j2qRzZU<RTSEe_;rKwgSWgt~r`$ew^mCkZK3FLUrfqiEy9O966b(;bz}{VC$R=+FYzMD3n~!Ek6AB|{!ND3@nm9w1P0mB{ebYvTsq@|Y=pwQMp3##6o;IwWd%V42*)EZhcQV;}FYTWrL_Nmf(VDp50_ep(Ur0TcsT=By`gqphe>!t`cg!KP!Dwx+K52%nUekX6FzSM)stD_T!_NpkVCH0#{qpx07co0Rf)%@lmL4Lyi)6nCft_5Xy=&BE`c^@^)`PxvKBhdmLx1pm_`9GkGvsckCchWO6V0F&DmtUq+e+26fHmB#B<O9>Yl~cQM1{#xtCV!_JiX7ez+b%cXaGzAn=L%*kfc>T5d~RWOEqTQiT&o+(MYFi*e&L?%ZVQg>^iIPdph-zo3Y@klH<`t{pBZhXP<42YfcQ+%*f><;*t;B+!*c7#nW?~%)Yav>fB=}$XMtvV<I&tuz-_hv_GBEUt`>IQ4mjhLbSX$^mq90ti-zY!)?%2%QRx=nXAzt2i&~%$!W8@YSjF3hYH+Y&9KLB>aJXFlkit*bD;_IWF+Q0qA2$^zyND<(eB+)@fD(#4gQo$cXRytSbbYG!K)(w8UtMm2ZQuu&q(EHC#^zs|6odZWzA^54(u7rCWS4{q-AFMsKzQXJ(g?Ldqx`Dr-xhCFzoZtK9e+gP7K=$uP6JfDE7d^E!8(t4bWCv-??vevdew!v?bAibp>?i+%B_Py3#XS{aTTZ$2qWu+4gHcVhj)_BKYIYhqu4Hee*iH{P6SN-~RC?`j7wq^yXu9`R;Y}@zdqs-cT8`N68;NSAljt7OGrx?byC?e{cWz((VE`Q_1Ci-^NFf5UZS;JX=vxu@bRdn&LdC;7fhBWRzZ)!%S2y%xC&_rvr?d;)K3TD$@fZ^x*c`VuY$2p7@G}%^<s2zS-V(fp>SfozZvaIMw{t#P*d!OJh8x-TwQ7_B^}Kn!Lsu?8vUyp7@KkciPCnElh+leI<+CDV%ee^c7K>3CSwX))*7Ez9d14T1}o=o{vmmqNmlGYZ|M_`3bMc{;e>C)4KEuP-_r8{NdII{X0r=mJeH>*3u7had930IM?{HyxO7|zNemHDyyUfT>x69QIr-*6q(X$U5hjH3Tv=KD1$z^3wBCZ(A`A4slYGu+8^EV5;4($S2DCV{OT$YKX5e=ebtu8VG!Z@l)i4$xqrTRq@RbFIzM4F*a4WeoeHZRI6Eh)#h8y6pFWj^*-tz~;=5@^AwEzt=b`<PV6;d-0h0Q*TLGj$2vT1_+aKnR+p30Nn;tfw>yAANMI)?QYwM@<g610ta3538mvoxCvHf9%zo+@Mq1Ekt3enfsGg;@U-PW%Cvj=x9ozvMeYF?JMx_fPu`aKh18u0Z`nd)DG;JO83YS_NOr=KFs>*;aWzOo;IE(ZUjidfVSuK{*YXZ6c;cOXXXAO8wPF27>Rc#2cKKg_0ne(h_S)ffe6V)f305Sp#-E4%21_O?Ij*%UimL}RRqN{i);&GXz72?jKvH>fn^H<b)KLd%C%xh8%v3MbbnLX#Iobaag(^axR;evX3O!{Gk_Rv_eQ"


def digest(data):
    return hashlib.sha256(data).hexdigest()


def save_json_exclusive(path, value):
    with path.open("x", encoding="utf-8", newline="\n") as stream:
        json.dump(value, stream, sort_keys=True, indent=2, allow_nan=False)
        stream.write("\n")


verifier_source = zlib.decompress(base64.b85decode(VERIFIER_B85))
assert digest(verifier_source) == VERIFIER_SHA256
namespace = {"__name__": "hung_compact_proof"}
exec(compile(verifier_source, "hung_kaggle_compact_proof.py", "exec"), namespace)

archive = pathlib.Path(
    "/kaggle/input/datasets/hungluongduc/hung-dev-shard0001-44b8532-20261027-v2/artifacts.zip.bin"
)
receipt_path = archive.with_name("receipt.json")
assert archive.is_file() and receipt_path.is_file(), "Exact saved dataset must be attached"
receipt = json.loads(receipt_path.read_bytes())
assert receipt["shard_id"] == SHARD_ID
assert receipt["artifact_bytes"] == ARCHIVE_BYTES
assert receipt["artifact_sha256"] == ARCHIVE_SHA256
assert receipt["plan_sha256"] == PROVIDER_PLAN_SHA256
assert receipt["shards_sha256"] == PROVIDER_PARTITION_SHA256
assert receipt["exit_code"] == receipt["local_verify_exit_code"] == 0
assert receipt["crc"] == "passed"

payload_matches = list(pathlib.Path("/kaggle/input").rglob("provider-payload.zip.bin"))
assert len(payload_matches) == 1, "Exact pinned payload dataset must be attached"
remote = payload_matches[0]
assert "hung-matched-payload-44b8532-20261027-v1" in str(remote)
assert namespace["file_sha256"](remote) == PAYLOAD_SHA256
with zipfile.ZipFile(remote) as payload:
    assert payload.testzip() is None
    parent_plan_bytes = payload.read("parent-plan.json")
    parent_partition_bytes = payload.read("parent-shards.json")
    portable = json.loads(payload.read("portable-plan.json"))
assert digest(parent_plan_bytes) == namespace["PARENT_PLAN_SHA"]
assert digest(parent_partition_bytes) == namespace["PARENT_PARTITION_SHA"]

repo_prefix = "/kaggle/working/hung-matched-shard0001-v1/repo"
for spec in [
    *portable["configs"],
    portable["selection"],
    *(job[role] for job in portable["jobs"] for role in ("source", "target") if job[role]),
]:
    spec["path"] = repo_prefix + "/" + spec["path"]
portable["input_base"] = repo_prefix + "/" + portable["input_base"]
portable["output_root"] = repo_prefix + "/" + portable["output_root"]
parent_partition = json.loads(parent_partition_bytes)
provider_partition = dict(parent_partition)
provider_partition["plan_sha256"] = PROVIDER_PLAN_SHA256

work = pathlib.Path("/kaggle/working/hung-compact-proof-shard0001-20261028-v2")
work.mkdir(exist_ok=False)
parent_plan_path = work / "parent-plan.json"
parent_partition_path = work / "parent-shards.json"
provider_plan_path = work / "provider-plan.json"
provider_partition_path = work / "provider-shards.json"
parent_plan_path.write_bytes(parent_plan_bytes)
parent_partition_path.write_bytes(parent_partition_bytes)
save_json_exclusive(provider_plan_path, portable)
save_json_exclusive(provider_partition_path, provider_partition)
assert namespace["file_sha256"](provider_plan_path) == PROVIDER_PLAN_SHA256
assert namespace["file_sha256"](provider_partition_path) == PROVIDER_PARTITION_SHA256

proof = namespace["verify_shard"](
    parent_plan_path,
    parent_partition_path,
    provider_plan_path,
    provider_partition_path,
    archive,
    SHARD_ID,
    ARCHIVE_SHA256,
    ARCHIVE_BYTES,
    SAVED_TRAINING_VERSION,
    INDEPENDENT_READBACK_VERSION,
)
assert len(proof["rows"]) == 20
proof["claimed_source_dataset"] = SOURCE_DATASET
proof["claimed_payload_dataset"] = PAYLOAD_DATASET
proof["verifier_sha256"] = VERIFIER_SHA256
proof_path = work / "compact-proof.json"
save_json_exclusive(proof_path, proof)
print(
    json.dumps(
        {
            "status": "VERIFIED_ARCHIVE_BYTES_AND_STAGE_PLAN",
            "shard": SHARD_ID,
            "jobs": len(proof["rows"]),
            "proof_bytes": proof_path.stat().st_size,
            "proof_sha256": namespace["file_sha256"](proof_path),
            "version_ids_independently_authenticated": False,
        },
        sort_keys=True,
    )
)
```
