```python
import base64, hashlib, json, pathlib, zipfile, zlib
import kagglehub

SHARD_ID = "shard_0001"
SOURCE_DATASET = "hungluongduc/hung-dev-shard0001-44b8532-20261027-v2/versions/1"
PAYLOAD_DATASET = "hungluongduc/hung-matched-payload-44b8532-20261027-v1/versions/1"
SAVED_TRAINING_VERSION = 362612528
INDEPENDENT_READBACK_VERSION = 362633606
PAYLOAD_SHA256 = "23a6d420523d990d07ff07d814ed642a851925283a102a3ae806f3b666e1d6f4"
ARCHIVE_SHA256 = "29e3db51106402942959f80762e4c27995ad4ec9f2253fd96cc8a4681d3b2dcd"
ARCHIVE_BYTES = 65931034
PROVIDER_PLAN_SHA256 = "80a1666cf2083594c9a2bc0b0a4dc3e30118eda43d07187cd5b6d8b51f90e0bb"
PROVIDER_PARTITION_SHA256 = "e2c6cb6f15a7473c6cdfd7912fefa838d63a770cf352dff93c2cccc3ae86a15c"
VERIFIER_SHA256 = "a7b6ebd0fbd69001e34b874fff0dba1c6d9ecc1e6dae682318a410074b537b90"
VERIFIER_B85 = "c-pmETXWks7Jk>Sz+xX(+R}<G%eJhkI^#{;soN&?#NNzowHgjZLJ}N`RCuxDdi>w_99#&H1ZAh4^@BwcI2Rmz=Yk8R^rpyIuqA22NFLm?h`eL>EKYY@mJ}r41X)CQQlumx`8J4SkA%>_U<YXuKaw5G4(`$=*=1?E8uog(8=jMUmhly58HrMslO!!jRwks#0-o?>?U5BL!VUQ)Sg&J7qM!&!nvtL^HXxCQz?+012sr2EZWnVF^?+l}(}Y~T%ZKE4qdSmonHOZ6mY^w66ZSO-3pI+gEOuo<f&|r%*p9)!qyUwd@Y>^rP*X6)E<qM<_&p;lP>^N2j3dWXFomJr^N3~t$O#NQS_a`~cnH%h8uloqz1}KIx5W2XWl?6#_X*$b(hMcT$b$kT<h`E$maTU|mNWgn3Gz+Mm-_WCPZRyT3ktjG558ORm`P#9;|}!Hf-aKBUhn;H?=FA!-@<<~BXm4oPEL+T=O^dU+39LIIy^l;noN#Pj?Y);=SN{M9-mK+&qimocYX2k^24ov{p-aC|K_I)BpyZM@gy3ZOvbCzlVz|vV~6MC@d-OTI$n-PljGx~@$zuOf>9W<$;nB0IypIH=d18&v<xO^j2WgEA8)U2uYRjIpGDE}_;5KH2Pa3tG7Qe6a5-X=<MH`nbU2xWhi6A;lhtY&Etluf<S06wj7BHX>FG%@Iy(=~R$%O2#8w0?;R7^BC#QY1(^Ok)cn!Y;a_|Gmi)`8>@E7ql1CY$Lt%r(kAX*=Iu@S_>6cEy<*^&-OkRv%bh2(&LD?n(tDU;6tCc+Dr_2Xc>jDo4^IYemmj}Av;@=t<41CJ~z9T*NqmtnaB(6heaC)LOpU?<V-8}?N`fs%5-Tee-YOb=Osb*e!z{GIQ9gkO&&!IlY?gbj*v7qdA)*~3qZ*VrTHe=uk}G{#ZFq<xV8|It1|`gb!DgDA^7uQeZ%d)-FN*q4%LK$s>GN57nsWtu`j+p;_l)|TNuKrob=@s)^QhNMymqFK(!rywrbWtOE`UvV+gg2+dQ?m}q9!(t8+J#=6-Irqp8?2mNAI1Qp)Pz&q%1^Zg`SrVoZIRA{6#p+-}2igL1Y&_>lUIa<V`hu4y#DK9^^z!SDg#|>$pKpHqK+@$MyiLgy@_(i^I|@oQ3<xv{gMp;Wyf6Cf;s6X+SKfosWreb7vq_{7pGU&HJqwuFQR=7v$2=FL3uO+J3f6dn-#O(GT}<(l02}@S)KHr_#eQ`0ENDU&$ASnSG!;HejN(H@C;C=|X|m#LNh%&IV(}))<)dLBX*3Hx3y5FTX+UNY%(j{^8T1_(KxR4|2V^zPQcZ-g8LBdu2^eMTRBUWLr?MYiAWp@z)EUKUvoBhjNg|=!5dz~q&jB^-H&LaO6!oN8Xn1g;k|duWF2sPK8RWNQ@W>24XN7nlh)jD(z9&_W?`LFm2==&20CPN2g3(T_pmk(J!tdX4VE_RO0Rs7stboalAjS}tL$LzwjITxKI=YeClQ8l0q<hFiAs*#{Hc%`e>cRgqcyu6Ml?BZf0dR~f62ed+T*PtufaOn;SD77@iPKP22jpP`u^afYoMrcz*P;H(9!VZ@tcO-S6d(ad9AdelqwQz*h-2^Lgu~!*JmOaj2%omB*rbsiVgV#_G{ZGa2{0Itzs;(FR3GaxSy}DJAcnG~B|J~dEMzFIfZ7Hp%}_5H&VfgQ732c|PE$7Ud3yHR@@0dqwd5o^P?@f%P+f4Ye||xE_2D+F0-PvTAANy572$-QKDwjdj@@3Zk9NiGj%su-u&a8#$WZ0u&h}ckm7kc3{Jy4@#@g3o`B<=giiN-8iZ!*bDrUjGYSKk<yR|dT0ggWjv+L1#SWz>+E5H-CLBdy%w{uyTL%hz|ih;R9$X!A&QJ6q|4{C!~g=ntIHFUVBmx%QV6_Ke?_yU5jMK=+K2^Oqd4htSwsWbwSsd*eCYSUMfW7oo#FilXGC}a$pXapa!tBS^;oVaw7)RlC=6b1!^wehuJJm>=MKH`@o*VTQs%nmY#G(Cw;Q_QIzI7X4het=YBaVd!b`z55bRdUE{O@UX@(`(hXPqt@)rmcZN3sx-=^TTwz<%L7EqunenEuG#&ub5!p{W{b{=OL^koI<S@8krpkAGaAWzuaJZ)6$B!OzG-HWp@3h6MPloWEIn@Ya3SbTCQX&S=&4HAe0LFRVz`wYc;Ab`_!;4<2lgdRHd?aO~R};aL<C0thFM(g5V0}6z(g?A=m3KtIo1HnLN9@^T?OGD!+@d%z?tHk<2%<d{Wvrt1ZDytXFQUf+^et$r^%xyCQ=1%blrDg*G8D2#q#kBIdh0S?trrqB%+PFQ%=woNW|X%QC|H2A*=p#23`1c=Tm6Pj+SD>)hpcl_P43LR0L(41F=L?tzO{%M-Lew+GyEbqmM}b#nN!FkCF=)ERYj-!JaIUX8y~NkOC)H2pgwlDp&}KfnS3^FA&SI_ef-v(vG)Y|y)RfC%dF2)kq&!s@u_4~95LVy|P?p$V|RP;BuXhKIY19g*J7cDA&ElfUd}32;>#=>ybXVnJN5pcc3)T$uft#vV{Ze2{D77$nLS(yZFq!FLqYZ-XjUdp`BWocvN>$NqkGt#+wi!X_@{Y9EA7X<=&FGGo_i&c9+&E4Q46JokfTp2lUt`hylV0)`HURIm{%<ZezFMDrF0ba-q>YB9W8k3k*%^xsJytZHL!!X5wqPE?UA*K@KbtTdA_E=B4t6RWbtNd(Zzr49+gknIZWgj>)eR1eL<3Q8Tx&6-7kYyntsOrq7D=VEUQqYNtl_T$?Mci_0DV?Dr#%3LL^r<x_uRKpxF04uOJbTB_!^j>dO+*=d$1x^Wqxli>n;0=|#1)>8>BGvo`;{Fh;JWIdl)_`jo&j>=+PF(}zd66ES%6UyMd{qKeJ5^-1FW>2{DZ7`<KVoyMnSgA=i~RzqsEu(%h&8rwQps>?@F6x}ZI%NUS$*%Z#%thAWi3xqG<tJLW|=!xsWhFo>ilvwyq{ifN_B9smnL4@_36(rd3yZKvbvqm>P{{e$3?JaRqoI29U<JzM6Y>v5a{WAxxd=}mnlf^&fEAD46Pbh3c!^Zpw<`(zvgr?<OSR2{YG(O6zc-qlW2Z+fX>i5VQ1J5I>_Xrv2i3}aU9h4^V^u~0AoYI%wZvng*W(8LBP(&YWnri+oSFpkkg;+0k!Bf@@91obE+EX;@ifp$b=&*FHD!tp6E$cZy{W5A>-rA6z<Yja8P^j?hY?=mL;XXw_cUu^QD5-IJ=zd4mYpn*?YxAqvK)L#vEmQ4?V6ifg-i<b31+Z<(bnt=ZPO;6;n&yAKyG_E$p=YYMhv3KjR(%Is_@09OpwTWw-A)vo!U4S#)&R*jC%k#XXo>F#K*tj#|CqbnQ!XiYF0TI_(Kvn(+-j(QgbgT4d!mX7-WWqB3YF=qpvb7`z1RU(>$^=a=oOJ-8m!YLVWF&#X);?^o4ZTf7VsCJ0spVtc>VO1(y_`Tq1g_{N#joJ8a+%3N(hh3R$|Gk`knoH0X*n_xIa?me?k2DL%Wt^EWQeY3udpH{=*wahlKpi8qyKMEdmI~J>LjKypD{Mq2%h0>2@VsOMGIkdlo+eY;@7w?O98dXtvPTO#_yla0tZ*gxHrnqb$20OfM)P~!W(<Zu5o??~kHXGI3eb7licJx%Ie&V2fuHQOlL##TKXtN0E=qrIE@ic9KsY`3zDARq#OIJ^WaNXU)d<Tsic+)=v%SNFw{i{`mm{PrXs+cya+xp_lu5Y`{X(Q3Ftpju~ZQEE4=jotU7y9(>fXop<ZNGL4#-KVS^wZ_Xs~@i}-}x6G-~M#<>C*q#e{L^t94`LsSZdiDQs`sVepGW|e{Ek#?Ao*wXTZiAX&!FaHb5meWwJi_EN&(qU<wGr;y`%Zfna!We`G1F&Mc;`0}Ip1KB%kg-aK4tcefX<b(-2qqD9zTF*l&>I975mk(+ex<ul#l`Hyd(6;0g-MSFu6t#67#pRVoFHi;~$rd=Djh9HaszQsjf;}jb;YJ74~IS#mi(DBuxH6L@ocKy`1@3|r$aW}!DM06v2_>o_q^&P-~C;i6tG;x_!?gGk-rNfJCT_OyxMN2<od6+?509|IjAElx1n`0flHP?57mF@v0?ZO@1i!R!pMEg@(7WxEnYM;r$K!!nt(<{f;rw0d=d?z;tGX`=))~7xm!yM3or-MZ_!)qQM-j-#v+V#8M9>&t!#i34(&t4)mm;(jkKSL_P>{gy9#eSU;#0Ls<4&7hLMnn1=I01Ps58IHT_(7I=zj*?)&WBA`BR|a4VUK8bIF#sNo%2}Nz-e{<3SoZcxW%Kj8`_7u?w0CUwb5KVhV5$WAl56)`qy^+vpcsUU9GVZ)SM=5G<U`*b;}Tp26p{}MS4XSmD^q37e@6+o@RyrnLXxe=D7%Avyq2KIO&6*1j$T9uL1ZX{jb^*RF{T9I=HjCRs9uU)V#Rrk%T?OJYh2=bzUk~SBD=mUa-DVR?j?u&`h<Zes6!xZRewoNwMoV494OLCmLlg!}G#XqEXSIv)VK24}~RIf3T<Kdu|*j;iY1}BEF9y&-cX&)%P(Y_&${elj))NKOt8x@B"


def digest(data):
    return hashlib.sha256(data).hexdigest()


def save_json_exclusive(path, value):
    with path.open("x", encoding="utf-8", newline="\n") as stream:
        json.dump(value, stream, sort_keys=True, indent=2, allow_nan=False)
        stream.write("\n")


verifier_source = zlib.decompress(base64.b85decode(VERIFIER_B85))
assert digest(verifier_source) == VERIFIER_SHA256
namespace = {"__name__": "hung_compact_proof"}
exec(compile(verifier_source, "hung_kaggle_compact_proof.py", "exec"), namespace)

archive = pathlib.Path("/kaggle/input/datasets/hungluongduc/hung-dev-shard0001-44b8532-20261027-v2/artifacts.zip.bin")
receipt_path = archive.with_name("receipt.json")
assert archive.is_file() and receipt_path.is_file(), "Exact saved dataset must be attached"
receipt = json.loads(receipt_path.read_bytes())
assert receipt["shard_id"] == SHARD_ID
assert receipt["artifact_bytes"] == ARCHIVE_BYTES
assert receipt["artifact_sha256"] == ARCHIVE_SHA256
assert receipt["plan_sha256"] == PROVIDER_PLAN_SHA256
assert receipt["shards_sha256"] == PROVIDER_PARTITION_SHA256
assert receipt["exit_code"] == receipt["local_verify_exit_code"] == 0
assert receipt["crc"] == "passed"

remote = pathlib.Path(kagglehub.dataset_download(PAYLOAD_DATASET, path="provider-payload.zip.bin", force_download=True))
if remote.is_dir():
    remote = remote / "provider-payload.zip.bin"
assert remote.is_file() and namespace["file_sha256"](remote) == PAYLOAD_SHA256
with zipfile.ZipFile(remote) as payload:
    assert payload.testzip() is None
    parent_plan_bytes = payload.read("parent-plan.json")
    parent_partition_bytes = payload.read("parent-shards.json")
    portable = json.loads(payload.read("portable-plan.json"))
assert digest(parent_plan_bytes) == namespace["PARENT_PLAN_SHA"]
assert digest(parent_partition_bytes) == namespace["PARENT_PARTITION_SHA"]

repo_prefix = "/kaggle/working/hung-matched-shard0001-v1/repo"
for spec in [*portable["configs"], portable["selection"], *(job[role] for job in portable["jobs"] for role in ("source", "target") if job[role])]:
    spec["path"] = repo_prefix + "/" + spec["path"]
portable["input_base"] = repo_prefix + "/" + portable["input_base"]
portable["output_root"] = repo_prefix + "/" + portable["output_root"]
parent_partition = json.loads(parent_partition_bytes)
provider_partition = dict(parent_partition)
provider_partition["plan_sha256"] = PROVIDER_PLAN_SHA256

work = pathlib.Path("/kaggle/working/hung-compact-proof-shard0001-20261028-v1")
work.mkdir(exist_ok=False)
parent_plan_path = work / "parent-plan.json"
parent_partition_path = work / "parent-shards.json"
provider_plan_path = work / "provider-plan.json"
provider_partition_path = work / "provider-shards.json"
parent_plan_path.write_bytes(parent_plan_bytes)
parent_partition_path.write_bytes(parent_partition_bytes)
save_json_exclusive(provider_plan_path, portable)
save_json_exclusive(provider_partition_path, provider_partition)
assert namespace["file_sha256"](provider_plan_path) == PROVIDER_PLAN_SHA256
assert namespace["file_sha256"](provider_partition_path) == PROVIDER_PARTITION_SHA256

proof = namespace["verify_shard"](parent_plan_path, parent_partition_path, provider_plan_path, provider_partition_path, archive, SHARD_ID, ARCHIVE_SHA256, ARCHIVE_BYTES, SAVED_TRAINING_VERSION, INDEPENDENT_READBACK_VERSION)
assert len(proof["rows"]) == 20
proof["claimed_source_dataset"] = SOURCE_DATASET
proof["claimed_payload_dataset"] = PAYLOAD_DATASET
proof["verifier_sha256"] = VERIFIER_SHA256
proof_path = work / "compact-proof.json"
save_json_exclusive(proof_path, proof)
print(json.dumps({"status": "VERIFIED_ARCHIVE_BYTES", "shard": SHARD_ID, "jobs": len(proof["rows"]), "proof_bytes": proof_path.stat().st_size, "proof_sha256": namespace["file_sha256"](proof_path), "version_ids_independently_authenticated": False}, sort_keys=True))
```
